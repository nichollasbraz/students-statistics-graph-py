```python
def classificar(row):
  if row["media_final"] >= 6 and row["faltas"] <= 10:
    return "aprovado"
  elif row["media_final"] > 6 and row["faltas"] > 10:
    return "reprovado por faltas"
  else:
    return "reprovado por nota"
```

```python
mainframe["media_final"] = round((mainframe["nota_p1"] + mainframe["nota_p2"] + mainframe["nota_p3"]) / 3)
```

```python
mainframe["media_final"] = round((mainframe["nota_p1"] + mainframe["nota_p2"] + mainframe["nota_p3"]) / 3)
print(f"média: {round(mainframe['media_final'].mean(), 2)}")
print(f"mediana: {round(mainframe['media_final'].median(), 2)}")
print(f"moda: {round(mainframe['media_final'].mode(), 2)}")
print(mainframe["media_final"].describe())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

for col in ['nota_p1', 'nota_p2', 'nota_p3']:
    mainframe[f'{col}_rand'] = np.clip(mainframe[col] + np.random.uniform(-1, 1, size=len(mainframe)), 0, 10)

mainframe['media_final_rand'] = round((mainframe['nota_p1_rand'] + mainframe['nota_p2_rand'] + mainframe['nota_p3_rand']) / 3)

plt.figure(figsize=(10, 6))
sns.histplot(mainframe['media_final_rand'], kde=True, bins=range(int(mainframe['media_final_rand'].min()), int(mainframe['media_final_rand'].max()) + 2))
plt.title('Distribuição da Média Final (Notas)')
plt.xlabel('Média Final')
plt.ylabel('Frequência')
plt.xticks(range(int(mainframe['media_final_rand'].min()), int(mainframe['media_final_rand'].max()) + 2))
plt.grid(axis='y', alpha=0.75)
plt.show()
```
